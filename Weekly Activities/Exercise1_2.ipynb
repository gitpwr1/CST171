**Exercise 1.2: Variable Operations**
Using your variables from Exercise 1.1:
1. Create a new variable that contains your full name and program of study as a
formatted string
2. Calculate and store your age in days (approximate)
3. Convert your height to feet (1 meter = 3.28084 feet)
4. Create a variable that stores the result of a comparison between your age and 21
Print all results with appropriate labels.

```python
name = input("What's your full name? ")
age = int(input("What is your AGE? "))
height = float(input("What is your height? Cm "))
program = input("What's your study program? ")
profile_info = f"Name: {name} | Program of Study: {program}"

age_in_days = age * 365.25

height_in_feet = height * 3.28084

is_older_than_21 = age > 21

print("--- Personal Profile Information ---")
print(profile_info)
print(f"Approximate age in days: {age_in_days:,.0f} days")
print(f"Height: {height_in_feet:.2f} feet")
print(f"Is older than 21? {is_older_than_21}")
```
