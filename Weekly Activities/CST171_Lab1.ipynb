**Part 2: Conditional Statements**

**Exercise 2.1: Basic If Statements**

Write a program that asks for a number and then tells the user if it is:


*   Positive, negative, or zero
*   Even or odd





```python
while True:
    try:
        user_num = int(input("Please enter an integer: "))
        if user_num > 0:
            print("The number is positive.")
        elif user_num < 0:
            print("The number is negative.")
        else:
            print("The number is zero.")
        break

    except ValueError:
        print("That was not an integer! Please try again.")

print("Thank you! You entered the number:", user_num)
```

**Exercise 2.2: Grade Calculator**
**Create a program that:**
1. Asks for a student's score (0-100)
2. Converts the score to a letter grade according to:


○ A: 90-100

○ B: 80-89

○ C: 70-79

○ D: 60-69

○ F: 0-59

3. Prints the score and corresponding letter grade
4. If the score is out of range (negative or > 100), display an error message

```python
while True:
    try:
        user_num = int(input("Enter Student's Score 0-100: "))

        if 0 <= user_num <= 100:
            print("Valid score accepted!")
            break
        else:
            print("Error: The number must be between 0 and 100. Try again.")
    except ValueError:
        print("That was not an integer! Please try again.")

if user_num >= 90:
    print("Your grade is A")
elif user_num >= 80:
    print("Your grade is B")
elif user_num >= 70:
    print("Your grade is C")
elif user_num >= 60:
    print("Your grade is D")
else:
    print("Your grade is F")
```
